```python
from tensorflow.keras.models import load_model #for loading the model that we have created
import pickle
import pandas as pd
import tensorflow as tf
```

```python
with open('label_encoder.pkl','rb') as file:
    label=pickle.load(file)
```

```python
with open('onehot_encoder.pkl','rb') as file:
    onehot=pickle.load(file)
```

```python
with open('scaler.pkl','rb') as file:
    scaler=pickle.load(file)
```

```python
model=load_model('model.h5')
```

```python
user={'Pclass':3,'Sex':'female','SibSp':1,'Parch':1,'Fare':11,'Embarked':'Queenstown'} #Enter user data to predict the survival.
```

```python
user=pd.DataFrame([user])
```

```python
user
```

```python
user['Sex']=label.transform(user['Sex']) #fit transform we use when we are first time encoding the data but here we are using the same process we used in training to just transform the user data.
```

```python
embarked=onehot.transform(user[['Embarked']])
```

```python
embarked=pd.DataFrame(embarked,columns=onehot.get_feature_names_out())
```

```python
user=pd.concat([user.drop(columns=['Embarked']),embarked],axis=1)
```

```python
user
```

```python
user[['Pclass','SibSp','Parch','Fare']]=scaler.transform(user[['Pclass','SibSp','Parch','Fare']])
```

```python
user
```

```python
y=model.predict(user) #model will predict probability because it is a classification problem.
```

```python
y=y[0][0]
```

```python
if y>0.5:
    print("The passenger will survive")
else:
    print("The passenger will not survive")
```
