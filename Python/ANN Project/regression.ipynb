# Train Test Split

```python
X=data.drop(['cognitive_load'],axis=1) #independent variables
Y=data['cognitive_load'] #dependent variable
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X,Y,test_size=0.2,random_state=42) #we will not use validation data as we have very less data here. We are only trying to understand regression here.
```

```python
X_train.shape
```

# Scaling

```python
from sklearn.preprocessing import StandardScaler
```

```python
scale=StandardScaler()
```

```python
X_train[['hours_coding','coffee_intake_mg','distractions','sleep_hours','commits','bugs_reported','ai_usage_hours']] = scale.fit_transform(X_train[['hours_coding','coffee_intake_mg','distractions','sleep_hours','commits','bugs_reported','ai_usage_hours']]) #we apply scaling onto continuous columns and not binary columns like task_success. We will not scale the target variable(cognitive_load) as it is a regression problem and we want to predict the actual value of cognitive_load.
```

```python
X_test[['hours_coding','coffee_intake_mg','distractions','sleep_hours','commits','bugs_reported','ai_usage_hours']] = scale.transform(X_test[['hours_coding','coffee_intake_mg','distractions','sleep_hours','commits','bugs_reported','ai_usage_hours']]) #we apply scaling onto continuous columns and not binary columns like task_success. We will not scale the target variable(cognitive_load) as it is a regression problem and we want to predict the actual value of cognitive_load. Here only transform is used as we have already fitted the scaler on training data and we want to use the same scaling parameters on test data.
```

# data leakage
If we scale the data before train test split then the mean of entire data is calculated which has both training and testing, so training data scaling is influenced by testing data also. which leads in leakage of information. We cannot figure whether the model has learned perfectly or was already influenced initially. Hence we split the data first and apply scaling to only training data(fit.transform) and onto test data we only apply the transformation as done in training data not fit. Basically as per the rules set in training data(mean value and std deviation), we apply the same onto test data. It is always a good practice to split data into train test first and then scaling. We scale not column by column but the entire dataset to get the mean value and std value of entire dataset and then subtracting every value from mean(of entire dataset)/std value(entire dataset), in this way only we can form a similar range between all values.

# Model Building

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.callbacks import EarlyStopping
```

```python
model=Sequential([Dense(64,input_shape=(X_train.shape[1],)),Dense(32,activation='linear'),Dense(1,activation='linear')]) #we will not create a complex network here because we have very less data and will result in overfitting. In regression problem we use linear activation function as we don't need to shrink any output value. we want continuous output values not probability scores. So not mentioning any activation function is synonymous to using linear(does not introduce non linearity to output like sigmoid,relu etc did) activation function. If we don't use activation finction like relu in the hidden layers then model will collpase in linear model. It can still learn or make predictions but cannot predict complex patterns. Like if house prices are increasing non linearly like small house price increase slowly,medium house rapidly and huge house  differently again, linear model will predict something like house size increases pice increases at a constant rate. So ofc depends upon the use case if you want to use relu in hidden layers or not.
```

```python
model.summary()
```

```python
model.compile(optimizer='adam',metrics=['mae'],loss='mse') #we use mean squared error as loss function for regression problems. We can also use mean absolute error as loss function. But MSE is more sensitive to outliers and we want to penalize the model more for large errors.
```

```python
model.fit(X_train,Y_train,epochs=100) #for less data here we can use 100 epochs
```

```python
from sklearn.metrics import mean_squared_error,mean_absolute_error,r2_score
```

```python
Y_predict=model.predict(X_test)
```

```python
mean_squared_error(Y_predict,Y_test) #In regression problems, there is no accuracy score like classification problems. So we use metrics like mean squared error, mean absolute error and r2 score to evaluate the model performance. It shows how far my predictions are from the actual values. The lower the value of MSE and MAE, the better the model performance. The higher the value of r2 score, the better the model performance. The best value of r2 score is 1 and worst is 0. It can also be negative if the model is performing worse than a horizontal line. With this high mse there could be overfitting but we just want to understand regression.
```
