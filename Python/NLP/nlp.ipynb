#### Feature Engineering, Modelling and Deployment


**1. Introduction to Feature Engineering and Vector Embeddings**
*   **The Core Concept:** Mathematical models (like Large Language Models) cannot understand human text natively. To achieve **Natural Language Understanding (NLU)**, text data must be converted into a numerical format.
*   **Vector Embeddings:** The process of converting individual words or tokens into corresponding series of numbers is called creating vector embeddings. For instance, a word like "flu" might be converted into a numerical vector like "1a 3 9a 8".
*   **Vector Databases:** The complete collection of vector embeddings from a text corpus is stored in a vector store or vector database. 
*   **Relevance to GenAI:** Understanding how text is converted into numbers is essential when building Generative AI (GenAI) applications and chatbots using APIs from OpenAI, Google Gemini, or open-source Hugging Face models like Mistral and LLaMA.

**2. Foundation Models and the Evolution of GPT**
*   **Foundation Models:** Companies like Google and OpenAI train massive "foundation models" on terabytes of internet data. Initially, these models are trained on a single, straightforward task: **next-word prediction**.
*   **GPT Architecture:** GPT stands for **Generative Pre-trained Transformer**. 
    *   *Generative:* Capable of generating new content.
    *   *Pre-trained:* Trained on a huge corpus of data.
    *   *Transformer:* The deep learning architecture powering the foundation model.
*   **Fine-Tuning:** A foundation model built for next-word prediction is fine-tuned (trained further) to perform specific tasks, such as behaving as a complex question-answering system. This transition is what turns a base GPT model into "ChatGPT".

**3. RAG (Retrieval-Augmented Generation) and Semantic Search**
*   **The Hallucination Problem:** A major flaw with LLMs is "hallucination"—the tendency of a model to confidently provide wrong answers when it lacks knowledge. 
*   **The RAG Solution:** To prevent hallucination (e.g., in a medical chatbot), developers can provide the model with a specific, trusted knowledge base (like a medical textbook). 
*   **Semantic Search:** When a user asks a question, their text is converted into a numerical vector. The application searches the vector database (the embedded knowledge base) to map the user's vector to the most relevant information based on exact semantic meaning. Tools like **FAISS** (Facebook Artificial Intelligence Semantic Search) are used for generating embeddings and conducting semantic search.

**4. Evolution of NLP Feature Engineering Techniques**
*   **Early/Basic Approaches:** Initial feature engineering manually counted the frequency of positive, negative, and neutral words to determine sentiment. However, this method completely fails to capture semantic meaning.
*   **Progression of Vector Techniques:** 
    1.  **One-Hot Encoding:** Early approach, superseded due to inefficiency.
    2.  **Bag of Words & N-Grams (Bigram/Trigram):** Incremental improvements that still struggled to grasp deep semantic context.
    3.  **TF-IDF (Term Frequency-Inverse Document Frequency):** Effective for simpler, heuristic (rule-based) chatbots.
    4.  **Word2Vec:** A deep learning neural network technique that successfully captures the exact semantic meaning of text data, making it optimal for modern NLP and GenAI applications.

**5. Machine Learning (ML) vs. Deep Learning (DL) in Feature Engineering**
*   **Machine Learning (ML):**
    *   Requires manual creation of numerical features based on extensive **domain knowledge**.
    *   **Advantage:** High **Interpretability**. You can easily explain to stakeholders *how* and *why* the model arrived at a specific prediction (e.g., "The model rated the text positive because it had 5 positive words and 2 negative words").
*   **Deep Learning (DL):**
    *   The DL algorithm automatically generates features from the raw data. No manual feature engineering or domain knowledge is required.
    *   **Disadvantage:** Complete lack of interpretability. It acts as a "black box," making it impossible to explain exactly how it reached its 98% accuracy.

**6. The Modeling Phase: Choosing the Right Architecture**
Applying a specific modeling technique depends heavily on the **nature of the problem** and the **amount of data available**.
*   **Heuristic (Rule-Based) Chatbots:**
    *   Operate on predetermined "if-then" rules or keyword triggers (similar to a decision tree). If a user asks about "operating hours" and "electronic store", the model triggers a pre-programmed response.
    *   **Pros:** Highly cost-effective, require very little data, and do not hallucinate (they give a direct answer or admit they do not know). 
    *   **Best for:** Standard FAQs and customer support.
    *   *Note:* Heuristics can be combined with Machine Learning (e.g., combining word-count sentiment with a rule checking if an email comes from a known spam ID) for better results.
*   **Generative AI (Deep Learning) Chatbots:**
    *   Utilize LLMs, Deep Learning, and **Reinforcement Learning** (a reward/penalty system for predictions) to continuously learn and adapt.
    *   **Pros:** Can understand complex inputs and generate creative, highly human-like responses.
    *   **Best for:** Complex problem-solving, creative writing (poems, stories), and highly conversational interfaces. Requires massive amounts of training data. 
    *   These heavily rely on **Transfer Learning**, where knowledge from a model pre-trained on a massive dataset is transferred to solve a new, specific use case.
*   **Cloud APIs:**
    *   Ready-made solutions (like Dialogflow) that solve NLP problems directly without building from scratch, though they come with usage costs.

**7. The Evaluation Stage**
Simply looking at a model's accuracy during training is insufficient. Evaluation is split into two categories:
*   **Intrinsic Evaluation (Pre-Deployment):** Measures the mathematical performance of the model using metrics like Accuracy, Precision, Recall, F1 Score, and Perplexity. 
    *   **Perplexity:** A crucial metric that measures how "confused" a model is when trying to predict the next word accurately.
*   **Extrinsic Evaluation (Post-Deployment):** Measures how well the model actually solves the customer's problem in the real world. For example, if a keyboard model predicts a next word but out of 100 users, no one clicks the suggestion, the extrinsic evaluation is extremely poor regardless of intrinsic accuracy.
*   *Key Takeaway:* Good extrinsic evaluation guarantees good intrinsic evaluation, but good intrinsic evaluation does *not* guarantee good extrinsic evaluation.

**8. Deployment, Monitoring, and Updation**
Once a model is thoroughly evaluated, it enters the final three stages:
1.  **Deployment:** Deploying the model to a cloud server. Depending on the use case, it can be deployed as a whole service (e.g., a WhatsApp chatbot) or as an API/microservice (e.g., an email spam classifier plugged into Gmail).
2.  **Monitoring:** Constantly tracking the model's key performance metrics via a dashboard to compare its current real-world performance against its past performance.
3.  **Updation:** Retraining the model with new data when monitoring reveals a drop in performance. For example, if a spammer creates a new email ID to bypass the spam classifier, the model must be updated with data containing this new ID to adapt.

**2. What is a Recommendation System?**
*   **The Concept:** A system designed to suggest items to a user based on specific algorithms and historical patterns. 
*   **Layman Example:** It acts like a perceptive shopkeeper; if a customer buys *Harry Potter* and enjoys the fantasy/magic genre, the shopkeeper (the recommendation system) will suggest other books with similar magical or dramatic themes.
*   **Real-World Applications:** Every major platform uses these systems, such as Amazon suggesting similar products, Netflix recommending movies, or Spotify suggesting songs by the same artist (e.g., Arijit Singh).

**3. The Four Types of Recommendation Systems**
Understanding these four categories is crucial for building and explaining recommendation models.

*   **1. Popularity-Based Recommendation System:**
    *   Suggests the most popular items to all users globally, regardless of individual taste.
    *   *Examples:* The IMDb Top 100 highest-rated movies (e.g., *The Shawshank Redemption*, *The Godfather*) or the YouTube "Trending" section.
    *   *Logic:* Uses internal formulas that factor in metrics like total views, the volume of likes, and positive comment sentiment.
*   **2. Content-Based Recommendation System (The Project Focus):**
    *   Suggests items based exclusively on the specific content the user is currently consuming.
    *   *Examples:* If you watch four thriller movies on Netflix, the engine will recommend another thriller. If you watch cricket or AI videos on YouTube, your feed will populate with similar content.
    *   *GenAI Link:* ChatGPT uses this logic via reinforcement learning with human feedback; when it asks which of two generated answers you prefer, it learns your "taste" and generates future answers in a similar pattern.
*   **3. Collaborative Filtering Recommendation System:**
    *   Recommends items based on the tastes and patterns of *other, similar users*.
    *   *Logic:* If User 1 and User 2 both give similar ratings to a thriller, a sports movie, and a comedy, the system identifies a taste overlap. If User 2 then highly rates a new movie (e.g., *Prisoners*), the system assumes User 1 will also like it and recommends it to them.
*   **4. Hybrid Recommendation System:**
    *   A complex combination of multiple systems, typically merging Content-Based and Collaborative Filtering.
    *   *Usage:* This highly complex approach is used by major tech giants (YouTube, Netflix, Amazon Prime), with each company utilizing its own proprietary internal formulas.

**4. The Mathematical Intuition Behind Data Retrieval**
*   **Multi-Dimensional Space:** Items (like movies or textual words) are mapped as numerical data points on a graph (either a 2D plane or a multi-dimensional space).
*   **Distance Calculation:** The system calculates the distance between these points; the data points with the minimum distance from the user's current choice are recommended.
*   **Euclidean Distance vs. Cosine Similarity:** While Euclidean distance works for simple 2D graphs, real-world systems (like Netflix's database or ChatGPT processing a 15-word prompt) map data across hundreds of thousands of dimensions. In these high-dimensional spaces, a metric called **Cosine Similarity** (or Cosine Distance) is used to mathematically identify the closest and most relevant data points.

````python
def stemming(text):
    L=[]
    for i in text.split(): #to convert string into list
        L.append(ps.stem(i)) #to stem every row in tags
    return " ".join(L) #to convert list back in string
````

## 2. Foundations of Git vs. GitHub

It is a common misconception that Git and GitHub are the exact same thing. They serve different but closely related purposes in version control and deployment.

### A. What is Git?
*   **Definition:** Git is a local **Version Control System (VCS)**.

## 5. Understanding Local vs. Remote & Git Statuses

### A. Local vs. Remote
*   **Local:** Refers to your personal computer, laptop, and the files sitting on your physical drive.
*   **Remote:** Refers to the online cloud platform (GitHub) where your code is permanently stored.

### B. Git Statuses
Git classifies files in your project directory under different states. You can inspect these states at any time by running:
```bash
git status
```

There are four primary file states:
1.  **Untracked (Represented by a `U` symbol next to the file in VS Code):**
    *   These are brand-new files created in your local directory (e.g., `index.html` or a new script) that Git has never seen before. GitHub and the local Git system are completely unaware of these files.
2.  **Modified (Represented by an `M` symbol next to the file in VS Code):**
    *   These are files that already exist in both your local directory and the remote GitHub repository, but you have made changes (added, edited, or deleted code) on your local machine that have not yet been updated in the cloud.
3.  **Staged:**
    *   Files that have been marked as "ready for commit" using the `git add` command. They are placed in an intermediate "staging area" before being locked in.
4.  **Unmodified / Unchanged:**
    *   Files that are completely in sync between your local computer and the remote repository. No new changes have been made to them.

## 6. Step-by-Step Git Commands Workflow (Local to GitHub)

When you are ready to save your progress and send your code from your local machine to GitHub, you must execute a strict sequence of commands:

### Command 1: Initialize Git
```bash
git init
```
*   **What it does:** Initializes Git inside your current project folder.
*   **Result:** It creates a hidden folder named `.git` in your directory.
*   **Purpose:** This hidden `.git` folder acts as the local version control engine. It immediately starts monitoring and tracking every single change (additions, modifications, and deletions) made to files within this folder.

### Command 2: Check the Current Status
```bash
git status
```
*   **What it does:** Displays the status of all files in the folder.
*   **Result:** It will highlight untracked files (like your streamlit `app.py`) in red, indicating they are not yet being tracked or prepared.

### Command 3: Stage the Files
You can stage files using two different syntaxes:
```bash
# Option A: Stage ALL files in the current folder (Recommended for full project push)
git add .

# Option B: Stage a specific file only
git add app.py
```
*   **What it does:** Moves the files from the "Untracked/Modified" state into the "Staging Area".
*   **Result:** If you run `git status` again, the files will appear in green under "Changes to be committed". This means they are prepared to be saved permanently.

### Command 4: Commit Your Changes
```bash
git commit -m "making changes permanent"
```
*   **What it does:** Takes a permanent "snapshot" or "screenshot" of your staged files.
*   **Syntax:** The `-m` flag stands for "message". You must write a descriptive message inside quotation marks summarizing the changes you made.
*   **Purpose:** The commit message allows team members and recruiters to look back at the history and easily understand what changes were made at that specific point in time. It turns temporary staged changes into a permanent local record.

### Command 5: Connect Local Git to Remote GitHub
```bash
git remote add origin https://github.com/your-username/movies-recommendation-system.git
```
*   **What it does:** Establishes a link connecting your local Git folder to your remote empty GitHub repository.
*   **Syntax:** 
    *   `remote add` tells Git to connect to an external cloud URL.
    *   `origin` is the nickname assigned to this specific GitHub repository. By default, developers use the name `origin` (though you could technically use any nickname, like `Amit`).
    *   The URL is the exact link copied from your newly created, empty GitHub repository page.

### Command 6: Rename/Overwrite the Branch
```bash
git branch -M main
```
*   **What it does:** Renames the default branch of your local repository to `main`.
*   **Context:** Historically, Git's default branch was named `master`. However, because `master` is considered a derogatory term, modern standard practice is to rename and overwrite it to `main`. The capital `-M` stands for "force rename and overwrite".

### Command 7: Push Your Code to the Cloud
```bash
git push -u origin main
```
*   **What it does:** Uploads all your locally committed files from your laptop to your remote GitHub repository (`origin`) under the `main` branch.
*   **Syntax:**
    *   `-u` stands for **Upstream**.
    *   By adding `-u`, you instruct Git to remember the connection between your local `main` branch and the remote `origin main` branch.
    *   **The Benefit:** In the future, whenever you make more changes and commit them, you do not need to type the full command. You can simply run `git push`, and Git will automatically know to send it to the `origin` repository under the `main` branch.

````python
# Emoji Handling
def remove_emojis(text):
    emoji_pattern = re.compile(
        "["
        u"\U0001F600-\U0001F64F"  # emoticons
        u"\U0001F300-\U0001F5FF"  # symbols & pictographs
        u"\U0001F680-\U0001F6FF"  # transport & map symbols
        u"\U0001F1E0-\U0001F1FF"  # flags
        "]+",
        flags=re.UNICODE
    )
    return emoji_pattern.sub(r'', text)
````

#### Basic text preprocessing:
1)Tokenization: Splitting raw text into meaningful tokens(word(separated from tab space), punctuation separated(I'm so I,',m),subword(3/4th part of the word(approx 4 letters) as one token like chatgpt),characters or sentence depending upon task) like chatgpt has a token limit depending upon the model gpt 3.5,4 etc so input+output= total tokens and if that exceeds the maximum limit of tokens in a model then the model tends to forget the initial conversation because the limit has been exceeded so it deletes previous memory hence text summarization upto the limit is a good practice so that we can enter it into our new chat. Splitting into tokens is required so that we can iterate over it and is helpful in next word prediction like in case of chatgpt. Tokenization is required in maintaining vocabulary size(No of unique words or tokens) and memory, in POS Tagging, Name Entity Recognition, Sentiment Analysis, LLM Models, basically wherever splitting is required.
# Main Types of Tokenizers:
1)Whitespace/simple Splitting: Splits on the basis of whitespaces(using split()), splits into words. It is quite fast. fails on the punctuation marks(like I'm) because they are not separated by whitespaces.
2)Regex based Tokenizers: Using Regular Expressions for splitting based on patterns like in punctuation marks(just like how we cleaned we can also separate it)
3)Rule based/Library Tokenizers: NLTK/Spacy/Standford NLP, useful in handling abbreviations/punctuation marks, used for word tokenizer/sentence tokenizer etc.
4)Character level: using simple for loop
5)Subword Tokenizer
# Limitations:
struggles to split on suffix(20km, we want 20, km not 20, k,m),prefix($20),infix(New-Delhi), special case rule(split a string into several tokens like U.S into U,.,S)
#### 2)Stemming and lemmatization:
Inflection is the modification of words to express different grammatical categories like tense, gender,voice, aspect, person, number, mood,case etc. In stemming(not proper root word as per dictionary) and lemmatization(proper spelling as per dictionary) we are removing inflection and bringing the word to their root word. Information Retrieval systems like google gives information about inflections of the root word as well(like if searched fish it gives info about fishnet, fishing, etc and vice versa because it stems all the inflections to it's root word and suggest info based on it) use NLTK library and stemmer(algorithm which is used for stemming purpose). For english language we use PorterStemmer and for other languages we use SnowballPorter. Stemming is quite fast because it uses an algorithm and because of this fastness it does not provide accurate root word. Lemmatization does not uses any algorithm so it is quite slow but give accurate results. It uses lexical dictionary to give proper words. Again use NLTK for this purpose. POS Tagging can be done by lemmatization(POS tagging is helpful in deriving meanings like bats root word is bat but if it is noun it is an object or animal but if verb then it is an activity)

### **Comprehensive Notes: Word2Vec Deep Dive (CBOW vs. Skip-Gram & Neural Network Foundations)**

---

#### **1. Introduction to Word2Vec Techniques & Embeddings**
* **The Goal:** To understand the different types of Word2Vec techniques and how word embeddings are generated. 
* **Prior Knowledge Recap:** We know that a Word2Vec model takes a vocabulary of words and generates a respective numerical embedding for each. 
* **Pre-trained Google Model:** 
    * Takes a word (e.g., "king") as an input.
    * Generates/predicts a corresponding dense vector embedding for that word.
* **Core Inquiry:** How does a deep learning model actually predict these embeddings based on a given word?
* **Two Main Word2Vec Techniques:** We can generate word embeddings using two distinct model architectures:
    1. **CBOW** (Continuous Bag of Words).
    2. **Skip-gram** (which is the exact reverse process of CBOW).

---

#### **2. Continuous Bag of Words (CBOW) and the "Dummy Problem"**
* **The Core Objective of CBOW:** We provide a word (or set of words) to predict the target word's embedding.
* **The "Dummy Problem" Concept:** 
    * In CBOW, we solve a **dummy problem** (also referred to as a "fake problem").
    * We do not solve this problem for the sake of the problem itself.
    * Instead, while the model is training to solve this fake problem, the **word embedding is generated as a byproduct** of the training process.
* **Example Text Corpus:** To understand the mechanics, we use a single sentence: *"hello welcome to my AI course"*.
    * *Note:* Real-world Word2Vec training uses massive datasets (such as a Harry Potter dataset containing entire books), but a single sentence is sufficient for conceptual demonstration.

---

#### **3. Window Size and Context Words in CBOW**
* **Step 1: Define the Window Size**.
    * **Window Size:** The number of words we consider in one go at a time.
    * Based on this size, we determine how many words will be categorized as "context words".
* **Window Size of 3 (Example):**
    * Means we look at a frame of 3 words from our sentence.
    * Out of these 3 words, some are labeled **context words** (used as inputs) and one is the target word to predict.
    * For the sentence segment *"hello welcome to"*:
        * **Context Words (Inputs):** *"hello"* and *"to"*.
        * **Target Word (Output):** *"welcome"*.
* **What is "Context"?** Context tells us if two words are similar. For example, in *"I like to pet cat"* and *"I like to pet dog"*, both *"cat"* and *"dog"* are used after the word *"pet"*, indicating they share a similar semantic context.
* **Continuous Frame Shifting:** CBOW is "continuous" because the window continuously shifts across the sentence to build training instances:
    * **Frame 1:** Inputs: *"hello"*, *"to"* $\rightarrow$ Target to predict: *"welcome"*.
    * **Frame 2:** Inputs: *"welcome"*, *"my"* $\rightarrow$ Target to predict: *"to"*.
    * **Frame 3:** Inputs: *"to"*, *"AI"* $\rightarrow$ Target to predict: *"my"*.
    * **Frame 4:** Inputs: *"my"*, *"course"* $\rightarrow$ Target to predict: *"AI"*.
* **Architecture:** To predict these target embeddings from context words, we use an **Artificial Neural Network (ANN)**.

---

#### **4. Deep Learning Foundations: Artificial Neural Networks (ANN)**
To understand CBOW and Skip-gram, we must understand basic deep learning concepts:
* **The Goal of Deep Learning:** To create an artificial replica of the human brain (the most intelligent thing on Earth).
* **The Biological Inspiration:** The human brain is a massive collection of connected physical neurons that transmit information.
* **Perceptrons:** In deep learning, these artificial brain neurons are called **perceptrons**.
* **Training Connection:** Advanced neural network architectures (like the ones built using perceptrons) are the exact structures used to train LLMs like ChatGPT.
* **CBOW Neural Flow:** We pass inputs through an input layer of perceptrons, mathematical calculations are processed in the hidden layer, and the final output layer perceptron generates the final embedding.

##### **A. The Mathematics of a Perceptron**
* Every artificial neuron (perceptron) is a mathematical formula, function, or algorithm.
* This underlying algorithm is **Linear Regression**.
* Just as Linear Regression uses $y = mx + c$, a perceptron uses:
$$\text{Output} = x_1w_1 + x_2w_2 + \text{bias}$$
*(where $x$ represents inputs, $w$ represents weights, and bias is the constant offset)*.

##### **B. Layers in a Neural Network**
A basic neural network contains three structural layers:
1. **Input Layer:** Receives the raw inputs ($x_1, x_2, \dots, x_n$).
2. **Hidden Layer:** Computes intermediate features.
3. **Output Layer:** Outputs the final prediction ($y$).
* **Fully Connected Structure:** Every perceptron in the input layer is physically connected to every perceptron in the hidden layer, which in turn connect to the output layer. These connections are called **channels**.

##### **C. Weights ($w$) and Biases ($b$)**
* **Weights ($w$):** Represent the **importance** of a specific input compared to other inputs.
    * *Cat vs. Dog Classifier Example:* If we input an animal's eye and nose, the nose is a much stronger distinguishing feature than the eye. Therefore, the channel connection for the "nose" input is assigned a significantly higher weight ($w$) than the connection for the "eye" input.
    * Important inputs = high weight; less important inputs = low weight.
* **Bias ($b$):** An adjustable value used to minimize the difference between actual and predicted values.
    * *Formula Adjustment Example:* If actual output $y = 10$, input $x_1 = 2$, and weight $w_1 = 3$:
    $$y = x_1w_1 + b \rightarrow 10 = (2 \times 3) + b \rightarrow 10 = 6 + b$$
    Here, the bias ($b$) must be adjusted to $4$ to achieve the exact actual output.
    * Bias can be thought of as a "cheating mechanism" because if the network's mathematical prediction is off, we change the bias value to force-align the output with the target.
* **ChatGPT Analogy:** ChatGPT uses context words as inputs. For the sentence *"I like to play cricket"*, the words *"I"*, *"like"*, *"to"*, and *"play"* are distributed across input neurons to predict *"cricket"*. If the prediction is wrong, the network's biases are adjusted.
* **Cost Function:** The metric that measures the difference between actual values and predicted values. Our goal in neural network training is to always **minimize the cost function**.

---

#### **5. Activation Functions**
An activation function is critical because it keeps outputs in a manageable, desired range (e.g., squashing an output of $100$ down to a range between $0$ and $1$).

##### **A. Threshold Function**
* Acts as a strict gatekeeper based on a threshold value (e.g., $10$).
* If the calculation $x_1w_1 + b_1$ is less than $10$ (e.g., $7$), the neuron does *not* activate. This means all channels stemming from this neuron will have zero effect on the final output.
* If the calculation is greater than $10$ (e.g., $14$), the neuron activates, and its connected channels impact the output.

##### **B. Sigmoid Function**
* Maps outputs strictly between $0$ and $1$. *(Not used in the CBOW technique)*.

##### **C. Softmax Function**
* Associated with the output layer of CBOW and Skip-gram networks.
* Primarily used for **multiclass classification** (predicting one of many output labels, unlike binary classification which only has two labels like $0$ or $1$).
* **How it works:** It converts a series of raw predicted values (e.g., $1, 4, 3, 6, 9, 12$) into corresponding **probability values** that lie between $0$ and $1$, ensuring that the sum of all predicted probabilities is exactly equal to $1$.

---

#### **6. Backpropagation and Loss Minimization**
* **Backpropagation:** The process where, if a prediction does not match the actual target output, we go backward through the network to adjust the weight and bias values.
* **Example:** If $x_1 = 2$, initial $w_1 = 3$, $b = 4$, our calculation yields $10$. If the actual target is $12$, backpropagation goes back and adjusts $w_1$ from $3$ to $4$ to achieve the correct prediction.
* **Loss Function:** Measures how far our prediction is from the actual value. We use backpropagation to continuously minimize this loss function so the model yields highly accurate decisions.

---

#### **8. The Skip-Gram Technique**
* **The Opposite of CBOW:** Skip-gram flips the input and output structure of CBOW.
* **Inputs vs. Outputs:**
    * **Input:** The target word (prediction word).
    * **Output:** The context words.
* **Shifting Window Example (Window Size = 3):**
    * **Frame 1:** Input: *"welcome"* $\rightarrow$ Outputs to predict: *"hello"* and *"to"*.
    * **Frame 2:** Input: *"to"* $\rightarrow$ Outputs to predict: *"welcome"* and *"my"*.
    * **Frame 3:** Input: *"my"* $\rightarrow$ Outputs to predict: *"to"* and *"AI"*.
    * **Frame 4:** Input: *"AI"* $\rightarrow$ Outputs to predict: *"my"* and *"course"*.
* **Neural Network Layout for Skip-Gram:**
    * **Input Layer:** 6 neurons (takes a single target word's one-hot vector).
    * **Hidden Layer:** 3 neurons (representing the window size/embedding dimension).
    * **Output Layer:** Two pairs of 6 neurons (12 total neurons, to output both context word vectors simultaneously).
* Just like CBOW, we apply Softmax, compute loss, backpropagate, and extract the hidden layer weights as the final dense embeddings.

---

#### **9. When to Use CBOW vs. Skip-Gram**
* **CBOW (Continuous Bag of Words):**
    * **When to use:** When your dataset is **small**.
    * **Advantage:** It is a much **faster** technique to train.
* **Skip-Gram:**
    * **When to use:** When you have a **large amount of data** (huge text corpus).
    * **Advantage:** It performs significantly better and yields higher quality embeddings when massive datasets are available.

---

#### **10. How to Improve Word2Vec Model Performance**
There are 3 primary parameters to adjust to optimize a Word2Vec model:
1. **Increase the Training Data:** Providing more data gives the neural network a much better mathematical mapping of context-to-target relationships.
2. **Increase the Vector Dimension:**
    * Google's pre-trained model uses **300-dimensional** vectors.
    * Advanced models like GPT/LLMs use **752** or thousands of dimensions.
    * High-dimensional spaces (e.g., 1000D) capture vastly superior semantic relationships compared to low-dimensional spaces.
3. **Increase the Window Size:**
    * If we increase the window size from 3 to 5 or 7, we capture more context words.
    * For a window size of 5, we look at 2 words to the left and 2 words to the right.
    * More context words give the model richer information, resulting in highly accurate embeddings.
    * *Trade-off:* Increasing window size directly **increases training time**.

````python
data=h1+h2+h3+h4+h5+h6+h7
````

````python
len(data.split()) #total 11 lakh words in harry potter corpus, split because we want words not characters otherwise length would come for characters like for loop.
````

````python
# Apply the NLP Preprocessing---1)Tokenozation--->Sentence Tokenixation,2)Stopword Removal 3)Lowercasing and the rest of basic preprocessing steps
from nltk import sent_tokenize
from gensim.utils import simple_preprocess
````

````python
potter=[]# because we are iterating over whole data and them appending the preprocessed data into an empty list
data=sent_tokenize(data)
````

````python
len(data) #63920 sentences(tokens)
````

````python
for i in data:
    potter.append(simple_preprocess(i))
````

````python
potter #outer list is the sentence and inner list is the words inside it. simple preprocess does basic cleaning like removing punctuations,numbers,tags and lowercasing, and give output in word token format.
````

````python
from nltk.corpus import stopwords
````

````python
stopwords=set(stopwords.words('english'))
````

````python
stopwords
````

````python
def stopword_removal(text):
    cleaned_text=[] #here sentence by sentence iteration is happening(potter has outer list(sentence))
    for sentence in text:
        new_sentence=[] #here word by word iteration is happening(potter has inner list(words inside a sentence))
        for word in sentence:
            if word.lower() not in stopwords: #words must be in lowercase and should not be in that stopwords list
                new_sentence.append(word)
        cleaned_text.append(new_sentence)
    return cleaned_text
    
    
````

````python
cleaned_potter=stopword_removal(potter)
````

````python
import gensim
````

````python
model=gensim.models.Word2Vec(window=10,min_count=2) #window is the window size that model takes as input in one go(context+predicted) and min count=2 is only those words are kept in vocabulary that are repeated twice or more. so for only those words embeddings are generated.(rare words are removed in large datasets because often they are spelling mistakes. for small we can go with min_count=1)
````

````python
model
````

````python
model.build_vocab(cleaned_potter) #we are building vocabulary(unique words) in corpus
````

````python
model.train(cleaned_potter, total_examples=model.corpus_count,epochs=model.epochs) #it trains the word2vec model on the data passed inside it, total_examples is total sentences present inside the model which can be found through model.corpus_count, epochs is how many times my model is going to iterate over the data again and again(for backpropagation or changing w,b values to reduce loss function)
````

## Principal Component Analysis(PCA):
