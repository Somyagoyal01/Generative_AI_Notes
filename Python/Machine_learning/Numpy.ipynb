## stacking and splitting in array-->size of the arrays must be the same.

## stacking--> joining the array together. row by row joining in case of of hz stacking and in case of vt, column by column joining. hstack() for hz joining and vstack() for vt joining.
splitting--->we are going to split arrays in 2 arrays. horizontal split(appears vt but it is horizontal because horizontal stacking is done so in this way splitting will also be done.),same goes with vertical split


```python
array_one = np.arange(1,13).reshape(3,4)

array_one
```

```python
array_two = np.arange(13,25).reshape(3,4)
array_two
```

```python
horizontal_array = np.hstack((array_one, array_two, array_one, array_two))
horizontal_array
```

```python
vertical_array = np.vstack((array_one, array_two))

vertical_array
```

## Horizontal split ---> np.hsplit(array, number of parts you want your array to be splitted at)

## vertical slit ---> np.vsplit(array, number of parts)

```python
np.hsplit(vertical_array, 2)
```

```python
np.vsplit(vertical_array, 2)
```

```python
np.vsplit(vertical_array, 3)
```
