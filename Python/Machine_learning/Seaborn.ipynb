Seaborn is a data visualization library that is built on top of matplotlib and is integrated with pandas.
# seaborn better than matplotlib?
1)syntax--->simpler,less lines of code is required. 
2)By default--->plots look classic that too without bringing any additional customization.
3)Plot types---->all of the visuals for statistical analysis(kde,box,regression)
4)Integration---->seaborn integrates with pandas dataframe but matplotlib with all datatypes(list etc)
5)More information can be extracted from the plots or visuals as compared to matplotlib,we can add on extra columns if we want to in seaborn.
6)A lot of new plots are supported in seaborn that are not supported in matplotlib.
# Two types of Graphs:
1)figure level graph: the container in which graph is created,that container is called figure.
The graph inside the container is called axis.
The figure and axis can both be referred to as an object just like xlabel,ylabel,ticks are also objects.
there is one figure and can be many axis.

1)Relational---->scatter plot and line plot. Relational is figure and scatter ,line are axis
2)Distribution---->Histogram,KDE,RugPlot
3)Categorical---->barplot,countplot,boxplot,violin,swarmplot
4)Matrix---->Heatmap,clustermap
5)Regression plot--->regplot
6)Multiplot--->jointplot,pairplot

## Facet Plot

#### Facet Plot can help us divide the plot as per the multiple categories of a categorical
column. Such as, let us say, we don't want to use `hue` and we want to divide the plot or graph
as per multiple categories of a categorical column, row-wise or column-wise, then we can go with
the facet plot.

#### Facet plot can help us separate the info Rowwise or Columnwise.

#### We can use Row or Column parameter to mention that if we want to separate the info
rowwise or columnwise.

#### Facet plot functionality is only possible in the Figure level plot.

#### As Figure level plot provides us the control over the whole figure and axis level plot only
provides us the control over the axis itself.

```python
sns.relplot(data=tips,x='total_bill',y='tip',col='sex',kind='scatter')
```

```python
#row-->smoker or not
#column--->male or female
sns.relplot(data=tips,x='total_bill',y='tip',col='sex',row='smoker',kind='scatter')
```

```python
## col_wrap ---> Lifeexpectancy for a country year wise in multiple columns

## Total countries ---> 12 ---> 12 Columns

### 12 columns ---> 1st row ---> 3 columns

## 2nd row ---> 3 columns ----> 4th row

## col_wrap ---> 3 ---> 3 total columns are going to be there in my first row
sns.relplot(
    data=gap,
    x='lifeExp',
    y='pop',
    col='year',
    col_wrap=3,
    kind='scatter'
)
```

#### KDE (Kernel Density Estimation) Plot ---> To show the distribution of the data inside a
Numerical Column itself, but over here instead of getting bins, we can get the smooth curve which
comes out as a probability density function representing the Distribution of the Data, as with the
curve how data is distributed exactly.(normal or skewed.)

#### Axis Level Function ---> sns.kdeplot()

#### ---> data ---> dataframe

#### ---> x ---> column whose distribution we want to find

#### ---> fill ---> To fill under the curve ---> True or False (By default ---> False)

```python
sns.kdeplot(data=tips,x='total_bill',hue='sex',fill=True)
#almost normal distribution
```

```python
sns.displot(data=gap,x='lifeExp',kind='kde')
```

#### Heatmap

#### In Every Classification Machine Learning Problem Statement, we use the Heatmap.

Heatmap uses color shades to show the size of Numbers in a Matrix 2D table.

#### Darker or Lighter colors indicate higher or Lower Values.

Super useful when we want to see correlation, Missing Data, comparison, density.

#### For ex ---> Sales over the Months in Different countries.
#### For coming up with a Heatmap or a Matrix which we can convert into a Heatmap, we create a
Pivot Table and then that Pivot Table only we can use to create a Heatmap.

#### pivot() ---> index, columns, values

gap.pivot(index='', columns='', values='')
#### In Matrix Plot, we are not having Figure level function but only Axis Level function

#### Heatmap ---> heatmap()

#### ---> data ---> matrix that has to be converted into a heatmap

#### ---> annot ---> It shows the numbers as well for the given rectangular grid

#### ---> cmap ---> A specific styling/theme for the heatmap itself ---> coolwarm, viridis

```python
dt=gap.pivot(index='country',columns='year',values='lifeExp')
```

```python
sns.heatmap(data=dt)
```

```python
dt=gap[gap['continent']=='Europe'].pivot(index='country',columns='year',values='lifeExp')
```

```python
plt.figure(figsize=(15,15))
sns.heatmap(dt,annot=True,cmap='viridis',linewidth='0.5')
```

#### Regression Plot ---> Regression Plot shows the Relationship between Two Numerical Variables

#### Line Plot ---> It can show us the future Trends.

#### Regression Plot ---> Scatter Plot + Line Plot

#### It checks that if the data follows a Pattern or Not.
(Linear Trend or Non Linear Trend)

#### ---> data ---> dataframe
