#### Hyperparameter Tuning--->GridSearchCV,RandomSearchCV
parameters passed inside algorithms like estimators etc, if we increase their value will we result in better accuracy?

```python
from sklearn.model_selection import GridSearchCV
from sklearn.ensemble import RandomForestRegressor

# Parameter Grid
param_grid = {
    'n_estimators': [100, 200],
    'max_depth': [4, 6, 8]
}

# Grid Search
grid = GridSearchCV(
    estimator=RandomForestRegressor(),
    param_grid=param_grid,
    cv=5
)

# Train
grid.fit(x_train, y_train)

# Best Parameters
print(grid.best_params_)

# Best Cross-Validation Score
print(grid.best_score_)
```

```python
from sklearn.model_selection import RandomizedSearchCV
from sklearn.ensemble import RandomForestRegressor

# Parameter Distribution
param_grid = {
    'n_estimators': [100, 200],
    'max_depth': [4, 6, 8]
}

# Randomized Search
random = RandomizedSearchCV(
    estimator=RandomForestRegressor(),
    param_distributions=param_grid,
    n_iter=10,
    cv=5
)

# Train
random.fit(x_train, y_train)

# Best Parameters
print(random.best_params_)

# Best Cross-Validation Score
print(random.best_score_)
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error

# Create Model
model = RandomForestRegressor(n_estimators=200)

# Train Model
model.fit(x_train, y_train)

# Predict
y_predict = model.predict(x_test)

# Mean Absolute Error
mae = mean_absolute_error(y_test, y_predict)

print("MAE:", mae)
#so basically here we use the tuned parameters from the grid search and random search to train the model and get the best results from our random forest model.
```
