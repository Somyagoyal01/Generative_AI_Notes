#### Feature Scaling
Basically we use distance based algo for classification(datapoints with near distance are classified in one group and far data points in another) and for that we calculate distance between these two clusters which has a formula like under root (x2-x1)^2+(y2-y1)^2.....etc so in that if x2 x1 are 53 21 so value will be (32)^2 but y2 y1 are 260000 and 1300000 so it will be (130000)^2 and adding both of them and calculating distance will be too out of range or actual distance. That's why we need to take them into same range to calculate correct distance between them. Feature scaling is basically scaling all the input column values in a similar range.
Types of feature scaling:
1)Standardization: scaling is done to keep all of the values in a standard scale range. for all scaled values,mean=0 and standard deviation=1(for all the new values that comes out after applying the formula,mean=0 so all new values will be scattered around origin). formula= value-mean of all values of that column/std deviation of all values of that column. this is optimal for distance based algo,if data is consisting of outliers
2)Normalization: values must be scaled down to normal range(between 0 and 1),formula=value-min value/max-min. it is used in neural networks(deep learning),where scale uniformity matters like Image data processing,don't use data with outliers as it won't give good results, graph will be of new datapoints all occuring withing a similar range of 0 to 1.

#### Machine Learning Algorithm(pattern)
This algorithm finds pattern(mathematical pattern) between input and output and predicts the final output.



1) Linear Regression:
It is a supervised ML ALgo that helps us predicting out continuous value. So basically when a graph would be plotted between input and output values we make a straight line which is close to all the actual datapoints but not overlapping everytime and this line is called best fitted line and based on that line when a new input is given it predicts the output by putting the respective point on that line. This is how linear regression prediction works. Distance between the actual and predicted point should be minimal. We don't want overfitting(100% accuracy of train data(overfitting),or line drawn through each datapoint or underfitting 40% or line drawn very far from data points we just want our model to be just perfect, line that is closer to all data points) we don't want our model to memorize everything perfectly in train data(that is 100% accuracy of train data we want our model to find patterns in train data so that it can predict any value. it is similar to mugging up vs actual understanding)
How to determine best fitted line?
Slope is change in output value with unit change in input value.(Determine m for test data)
Y-intersept--->what is output value when input value is 0.(Again c determine for test data)
Now with line equation y=mx+c we can determine output value by putting input values.
1)Simple Linear Regression-->one input predicts one output
2)Multiple Linear Regression--->Multiple inputs(multiple features) predict one output(y=c+m1x1+m2x2+m3x3...mnxn),x is the number of inputs.
To get the best fitted line we have to minimize(not zero but minimize) the distance(distance formula) between actual output and predicted output.
Assumptions for linear Regression:
1)Relationship between x and y should be linear.
2)input data should be independent from each other.
Limitation:
1)sensitive to outliers(would not give the right best fitted line)
2)cannot model non linear relationships(with increase in input output should also increase or slight decrease to achieve a linear model and vice versa)
3)Assumes linearity(as it is a linear based model but everytime we don't see linearity in our data cases)






2)Logistic Regression:
It is a supervised ML ALgo that is used for binary problem statement(0 and 1,two categories,classification is also multinomial(predicting multiple categories)) and it predicts the probability of occurance of a category.
If we use linear regression for such datasets then even if input va;ues are continuous output values are still in 0's and 1's. so even if we draw a line that is close to the datapoints, with new point it is always going to be greater or less than 1, not necessarily exact 1, and that is not accepted. so what we can introduce is decision boundary in which there is also a 0.5 between 1 and 0 and if output as per regression comes as greater than 0.5 then outcome it is most probable as 1 and if less than 0.5 then 0(not necessarily 0.5, but threshold/decision boundary could be any value between 0 and 1 depending on the use case). But in case of outliers, the output might exceed 1 then what can be done in that case? So we use squashing. basically draw line from 1 on y axis that will be parallel to x axis and from 0 also extend line, so now it becomes two lines and between that a slanted linear line(of decision boundary or linear regression-best fitted line(hence used the word regression here also)) so that limits the outcome probabilty to 0 and 1 only not exceeding 0(negative numbers) and not exceeding 1. #### Metrices in Regression:
Distance in between actual vs predicted point decides the best fitted line. So for this we draw many lines and see for which we get better metric value(less error).
Error(distance between actual vs predicted point),error more then not the best fitted line.
We need to measure how well our model can predict continuous target values so we use these metrices:
1)Mean Squared Error: Average of all error values( mean of all the distance between actual and predicted value of all continuous points), formula is 1/n summation of (actual-predicted)^2. Summation because many continuous values are there. n means total number of observations. As error is getting squared, so it shows a large value so it can be used when we want to penalize large errors for more severely use cases like finance,healthcare etc where errors matters a lot. units also get squared like cm^2 etc.
2)Mean Absolute Error: formula is 1/n summation of (actual-predicted). It is used when all errors are equally important and we want a robust metric system for outliers(in MSE,outliers would have devastated the prediction because it's value would have gotten squared and we could'nt find the best fitted line).
3)Root Mean Squared Error: under root (MSE)
Preferred when large errors are particularly undesirable(if individual squared errors are quite large so by taking root we minimize it's value so that it does not impact the overall system), sensitive to outliers.

Underfitting: When accuracy of training data is also less and with test data also less. Basically model has not learnt the patterns well.
Reasons: Model is too simple(linear model for non linear data), wrong features selected, less training time.
Bias: Error we get during the time of training. Underfitting is high bias(high error in training)
Variance: Error we get during the time of testing. Underfitting is high variance(high error in testing)
Overfitting: High accuracy during training(100%) but with test it is less. Model has mugged up things but didn't learn the underlying pattern. Overfitting is low bias and high variance.
Reason is too small dataset, too many features, too complex model.
Generalized Model: Training and test both high accuracy. Error minimized. low bias low variance.

#### Support Vector Machine
It is a supervised ML ALgo that is used for both regression and classification which aims to find the optimal hyperplane that best separates datapoints of different classes with maximum margin.
Hyperplane is a decision boundary which is capable to separate multiple classes from each other. In 2d we can have line to separate(in 2d hyperplane is line) and in 3d we have plane(hyperplane is plane) and 4d we have hyperplane(because 4d is not recognizable through human mind so we just call it a hyperplane). Along with hyperplane we also have marginal planes(in case of 2d marginal lines) which basically mean the plane that passes to the datapoints of each category that are closest to the hyperplane.(if 2 categories then 2 marginal planes) And these datapoints that are closest to the hyperplane are called support vectors. we have to choose the hyperplane in which distance between multiple marginal planes is maximum. if marginal planes are very closest to each other then they are not efficiently separated so for separation distance between marginal planes should be maximum. Hard Marginal plane is the plane when every datapoint is perfectly classified or separated through hyperplane and soft marginal plane is when some datapoints are missclassified through hyperplane(more noticable because datapoints are mostly random in order).
In coding, SVC(Support Vector Classifer) and SVR(Support Vector Regressor) are used for each classification and Regression. Parameters used are:
1)Kernel: used when data is not linearly separable. value of kernel decides what kind of data we are handling. Kernel='linear' mean data is linearly separable.(multiple categories could be separated by a straight line,mostly for binary output(two categories) so ex: when increase in marks it is pass and lowest fail, so it is easily linearly separabe). Kernel='Polynomial' mean when there are more than 2 output categories(suppose topper,pass,fail) so now data cannot be separated by just a straignt line so here curve is used for separation of data. Kernel='rbf' mean when there is a complex relationship between datapoints and they cannot be separated using a line or curve either then this kernel has the ability to transform this data into higher dimension and then probably we could separate the datapoints(suppose datapoints in inner circle and outer circle and some datapoints of outer circle got mixed with inner, so curve and line both cannot be used). Kernel='Sigmoid' mean there is a S-shaped curve formed in datapoints.(like data in low high low format)
2)Regularization(c) parameter: it's value can help us out with creating or sharpening our decision boundary and also help us out with avoiding overfitting or underfitting issues.
c value can help us separate every datapoints in category as every datapoint separation will lead to overfitting and some if left or misclassified will lead to underfitting.if c is very high then it perfectly classifies every datapoint and can form a hyperplane that gives 100% accuracy of training set but less of test that is why it can result in overfitting kind of issues. if c is less then it can misclassify a large number of datapoints and that is why it could lead to underfitting kind of issues. hence c should be moderate enough so that it could give better results. Range is from (10^-3 to 10^3 but always go with 1,10,0.1 values to avoid overfitting and  underfitting)
3)Gamma: Gamma focuses upon controlling the influence of each datapoint such that it creates the decision boundary around each datapoint. so high value of gamma leads to overfitting and low value leads to underfitting.

#### Decision Tree
It is a supervised ML Algo used for both classification and regression which models decisions as consequences and can be associated with it. It is a collection of nodes(root nodes are first conditions passed like age=18 and wrt ML ALgo it is the entire dataset at first from where the splitting starts, and leaf/terminal nodes are when the condition(features,input columns) is met true and some conclusion is made like adult and decision nodes are when condition is not met true and some another condition is passed like age>18 or node where splitting happens based on a feature when condition of root node is not true so we put another condition until we achieve leaf node(predicted class). Branches are this continuous flow of nodes(branch connect one node to another), these yes/no that decides the leaf and decision nodes. Prunning is a way in which we can remove unnecessary branches and it helps in reducing overfitting(more unecessary features or conditions then model cannot notice pattern and mugs up so result in overfitting so selection of important features are required,minimum depth(less branches)),pure nodes are the final leaf nodes(final predictions or conclusions)) that makes decisions if certain conditions are met(like the flowchart of if else clause). To validate the feature that is chosen is optimal or not we use some metrices like gini impurity, entropy and information gain. As we go lower in the tree hierarchy confusion of getting yes or no(branches) has to be reduced. Pure split is if a node(or a collection of nodes) is reaching final leaf node(pure node) after splitting(Yes/no) then that split is referred to as pure split. 
1)Entropy: It provides us the insight about the disorder in data splitting or how messy your data is after the split(how confusing your split becomes, as confusion should reduce to reach the final output)p+ is the probability of getting output through yes and p- is through no.
H(S) = -(p+ × log₂(p+)) - (p− × log₂(p−)), H=entropy. By using this formula we can validate if the node is leaf node or not. if entropy=0 after putting all the yes no probabilities of a node then it is a leaf node as entropy=0 means disorderness=0, if entropy=1 then it is an impure split and the node is not leaf node. it's value always lies between 0 and 1. less entropy value then more pure split so our goal is to reach less entropy value.
2)Gini Impurity: It measures the probability of a wrong classification. pi is p+/p-,Gini = 1 - Σ (pi)²,for binary classification, gini=1-(p+^2+p-^2). Range of gini inpurity lies between 0 and 0.5. so gini=0 mean pure split and gini=0.5 is impure split. gini=0 mean the probability of a wrong classification is 0 so it is rightly classified. for large datasets go for gini impurity.
3)Information gain: tells about the flow of input features that with what flow(f1 then f2 then f3 or f1 then f3 then f2) we could reach the output at earliest.
Information Gain(S, f) = Entropy(S)-Σ (|Sv| / |S|) × Entropy(Sv)
IG = Entropy(Parent) − Weighted Entropy(Children)
Information Gain (IG) = Entropy(Parent Node) − Σ[(No. of samples in Child Node / No. of samples in Parent Node) × Entropy(Child Node)]
Suppose if f1 is chosen as a root node and IG is 0.9 and if f2 is chosen as root node and IG is 0.1 then IG(f1)>IF(f2), so f1 should be chosen as root node. So higher IG lesser entropy better model. the entropy(confusion or disorderness) of child node must be lesser than parent node. so if it happens then difference in both will be higher so higher IG.

#### DBSCAN Clustering:
DBSCAN: Density based spatial clustering of Application with Noise. It cluster datapoints based on the density(datapoints which falls under same radius forms a cluster). It can find arbitary shaped clusters and identify outliers(datapoints which are outside that cluster radius). First find the radius(represented as epsilon(E)), min pts are minimum datapoints required to form a dense region. If the minimum points(or datapoints in addition to minimum points) are within that radius then it becomes core points. core points forms a cluster. Points which are outside that radius or not reachable from any core point is called Noise(outlier). A Border Point is a point that does not have enough neighboring points to become a Core Point (i.e., it does not satisfy MinPts). However, it lies within the ε-radius of a Core Point, so it is included in that Core Point's cluster. A Border Point belongs to a cluster but cannot expand the cluster further.A Noise Point is a point that does not have enough neighboring points to become a Core Point and does not lie within the ε-radius of any Core Point. Therefore, it does not belong to any cluster and is treated as an outlier.use cases are geospatial models,anamoly detection, astronomy etc.
