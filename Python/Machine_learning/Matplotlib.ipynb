## Univariate and Bivariate Analysis
Univariate---->Visualization is created/analysed for a given unique column(how data is spread in a single column)
Bivariate---->analysis is performed on two columns.It can be done with 1)Numeric-Numeric 2)Numeric-categorical 3)categorical-categorical
Multivariate---->work with multiple columns(more than 2)

plt.figure()---->takes parameter as fixed size(l,b value) and that determines the figure dimension of graph. ex:plt.figure(figsize=(10,6)) #determines how much area a graph should cover.
plt.title()---->It takes the value as the title of the plot as what x vs y is representing or what insight we are getting through this graph. as a parameter it can take colour as the colour of the title.
plt.ylabel()---->represents what y axis is representing to us.
plt.xlabel()----->represents what x axis is representing to us.
plt.grid()----->helps to bring grid on our visualization.
plt.show()----->helps in representing the chart
plt.xlim()----> it puts a limitation that how much the data must be represented on x-axis.
plt.ylim()-----> it puts a limitation that how much the data must be represented on y-axis.
## all these cannot be applied to pandas but in matplotlib we have a lot of customization.

# customization for line
parameters:
1)color---->what is going to be the color of the line by passing on color name or hashcodes.
2)marker---->how point of intersection appears.o,^,s,*,+,d,h,p(passed inside '')
3)linestyle---->defines the style of the line.'--','-.','-',':'
4)linewidth---->it shows the width of the line. 1,1.5,2etc
5)markersize---->change the size of the marker,default-->6
6)label---->labels what this specific line is representing,helpful while plotting multiple lines in a single visualization.

## Multiple Bar Charts

```python
sales.head()
```

```python
m=sales[sales['Product']=='Smooth Sliky Salty']
p=sales[sales['Product']=='50% Dark Bites']
d=sales[sales['Product']=='Eclairs']
```

```python
m=m.groupby('Date')['Amount'].sum()
p=p.groupby('Date')['Amount'].sum()
d=d.groupby('Date')['Amount'].sum()
```

```python
m
```

```python
m.index=pd.CategoricalIndex(m.index,categories=['January','February','March','April','May','June','July','August','September'],ordered=True)
m=m.sort_index()
p.index=pd.CategoricalIndex(p.index,categories=['January','February','March','April','May','June','July','August','September'],ordered=True)
p=p.sort_index()
d.index=pd.CategoricalIndex(d.index,categories=['January','February','March','April','May','June','July','August','September'],ordered=True)
d=d.sort_index()
```

```python
plt.figure(figsize=(10,6))
plt.title('Analysis of Product sales month on month',color='Red')
plt.ylabel('Total Sales')
plt.xlabel('Month Name')
plt.bar(m.index,m.values,color='green',label='Smooth Sliky Salty')
plt.bar(p.index,p.values,color='blue',label='50% Dark Bites')
plt.bar(d.index,d.values,color='red',label='Eclairs')
plt.legend()
plt.show()
```

xticks()---->This function helps us out with replacing original label with the new label as per our needs. xticks(original_label,new_label)

```python
import numpy as np
x=['January','February','March','April','May','June','July','August']
plt.figure(figsize=(10,6))
plt.title('Analysis of Product sales month on month',color='Red')
plt.ylabel('Total Sales')
plt.xlabel('Month Name')
plt.bar(np.arange(m.shape[0])-0.2,m.values,color='green',label='Smooth Sliky Salty',width=0.2)
plt.bar(np.arange(p.shape[0]),p.values,color='blue',label='50% Dark Bites',width=0.2)
plt.bar(np.arange(d.shape[0])+0.2,d.values,color='red',label='Eclairs',width=0.2)
plt.xticks(np.arange(m.shape[0]),x)
plt.legend()
plt.show()
#For equal spacing in bars
```

### Stacked Bar chart
pass bottom as a parameter in bar(),pass y values in bottom


```python
plt.figure(figsize=(10,6))
plt.title('Analysis of Product sales month on month',color='Red')
plt.ylabel('Total Sales')
plt.xlabel('Month Name')
plt.bar(m.index,m.values,color='green',label='Smooth Sliky Salty')
plt.bar(p.index,p.values,color='blue',label='50% Dark Bites',bottom=m.values)
plt.bar(d.index,d.values,color='red',label='Eclairs',bottom=(m.values+p.values))
plt.legend()
plt.show()
#length of bars shows maximum sales.maximum length means maximum sales of that.
```

## Pie Chart 
---> Pie Chart is being used for Bivariate Analysis and is being used for relating
Categorical-Numerical columns But in Pie Chart, we will look to the Percentage Distribution
(Univariate Analysis) and Bivariate Analysis (Categorical-Numerical) as in Terms of Category by
Category Percentage Distribution in Aggregation of Numeric Data

```python
data=df['Year_of_Study'].value_counts()
```

```python
plt.pie(data,labels=data.index,autopct='%0.1f%%',explode=[0.1,0,0,0,0],shadow=True)
plt.legend()
plt.show()
# Univariate Analysis
```

```python
# Bivariate Analysis
data=sales.groupby('Country')['Amount'].sum()
plt.pie(data,labels=data.index,autopct='%0.1f%%')
plt.title('Percentage contribution of countries in total sales')
plt.legend()
plt.show()
```
